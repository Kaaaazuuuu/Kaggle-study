データの場所を確認する

```python
from pathlib import Path

import numpy as np
import pandas as pd

print("現在の作業場所：",Path.cwd())
print("pandas：",pd.__version__)
```

データフォルダの場所を指定する

```python
current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_root = current_dir.parent
else:
    project_root = current_dir

data_dir = project_root / "data" / "titanic"

print("プロジェクトの場所:", project_root)
print("データの場所:", data_dir)
print("データファイル:", list(data_dir.iterdir()))
```

csvファイルを読み込む

```python
train = pd.read_csv(data_dir/"train.csv")
test = pd.read_csv(data_dir/"test.csv")
sample_submission = pd.read_csv(data_dir/"gender_submission.csv")
```

```python
print("train：", train.shape)
print("test：", test.shape)
print("sample_submission：", sample_submission.shape)
```

学習データとテストデータの違いを確認

```python
print("trainにだけある列:", set(train.columns) - set(test.columns))
print("testにだけある列:", set(test.columns) - set(train.columns))
```

各列の型と欠損値を確認

```python
column_summary = pd.DataFrame(
    {
        "data_type": train.dtypes.astype(str),
        "missing_count": train.isna().sum(),
        "missing_rate": train.isna().mean().round(3),
        "unique_count": train.nunique(),
    }
)

column_summary
```

生存者の割合を確認する

```python
survival_summary = train["Survived"].value_counts().sort_index()
survival_summary
```

```python
# 割合
train["Survived"].value_counts(normalize=True).sort_index()
```

```python
# 性別ごと
train.groupby("Sex")["Survived"].mean()
```

```python
# 客室クラスごと
train.groupby("Pclass")["Survived"].mean()
```
